# 1. 추정매출 5년 합치기 ✅
2021~2025 추정매출 5개 파일을 세로로 합치고 커피-음료만 남김
- 행 수 합 일치, 칸 55개
- 업종 1개, 분기 20개(20211~20254)
- 결과: 21,572행 → interim/cafe_sales_2021_2025.csv

```python
# 판다스 호출
import pandas as pd
```

```python
# 추정매출 21년 자료 불러오기 / 열, 행수 확인
sales21 = pd.read_csv("../data/raw/서울시_상권분석서비스(추정매출-상권)_2021년.zip", encoding="cp949")
sales21.shape
```

```python
# glob 호출 / 추정매출 자료 한 번에 불러오기
import glob
sales_files = sorted(glob.glob("../data/raw/서울시*상권분석서비스(추정매출-상권)_****년.zip"))
```

```python
# 불러온 데이터들 리스트 만들기 / 세로로 합쳐서 행열 확인 / 행수 합 일치 확인
result = []
for f in sales_files:
    result.append(pd.read_csv(f, encoding="cp949"))

sales = pd.concat(result, ignore_index=True)
print(sales.shape)

total = 0
for x in result:
    total = total + x.shape[0]
total == sales.shape[0]
```

```python
# 열 목록
sales.columns
```

```python
# 서비스_업종_코드_명에서 카페 업종 코드 찾기
sales["서비스_업종_코드_명"].unique()
```

```python
# 업종코드 커피인 데이터만 추출
cafe = sales[ sales["서비스_업종_코드_명"] == "커피-음료"]
cafe.shape
```

```python
# 추출한 데이터 업종코드 커피만 있는지 확인
cafe["서비스_업종_코드_명"].unique()
```

```python
# 21년1분기~25년4분기까지 빠짐없이 있는지 확인
sorted(cafe["기준_년분기_코드"].unique())
```

```python
# interim 폴더에 저장
cafe.to_csv("../data/interim/cafe_sales_2021_2025.csv", index=False, encoding="utf-8-sig")
```

# 2. 점포 합치기 ✅
2021~2025 점포 5개 파일을 세로로 합치고 커피-음료만 남김 / 25년 영어 칸 이름 → 24년 이름으로 교체
- 행 수 합 일치, 칸 14개
- 업종 1개, 분기 20개(20211~20254)
- 결과: 30,228행 → interim/cafe_store_2021_2025.csv

```python
# glob으로 점포 파일명 모으고 5개 맞는지 확인
store_files = sorted(glob.glob("../data/raw/서울시*상권분석서비스(점포-상권)_****년.zip"))
len(store_files)
```

```python
# 모은 파일명으로 파일 리스트에 담기 / 각 파일 열수 확인
store_list = [pd.read_csv(f,encoding="cp949") for f in store_files]

for x in store_list:
    print(len(x.columns))
```

```python
# (25년 파일만 칸 이름이 영어로 되어있는 문제) 24년 칸 이름 확인
store_list[0].columns
```

```python
# 25년 칸 이름 확인 (영어)
store_list[1].columns
```

```python
# (25년과 24년 칸 내용, 순서가 같은지) 24년 데이터 확인
store_list[0].head()
```

```python
# 25년 데이터 확인, 24년과 같음
store_list[1].head()
```

```python
# 25년 칸 이름을 24년 거로 대체 및 확인
store_list[1].columns = store_list[0].columns
store_list[1].columns
```

```python
# 5년 데이터 세로로 합치기 / 각 행수 합과 합본의 행수가 같은지 확인
store = pd.concat(store_list, ignore_index=True)

totall = 0
for i in store_list:
    totall = totall + i.shape[0]

totall == store.shape[0]
```

```python
# 칸 수가 14인지 확인
for i in store_list:
    print(len(i.columns))
print(len(store.columns))
```

```python
# 칸 제목 확인
store.columns
```

```python
# 업종 중 카페에 해당하는 이름 찾기
store["서비스_업종_코드_명"].unique()
```

```python
# 업종이 카페인 것만 추출 / 추출된 데이터의 업종 목록 확인
cafe_store = store[store["서비스_업종_코드_명"] == "커피-음료"]
cafe_store["서비스_업종_코드_명"].unique()
```

```python
# 추출된 데이터 행열수, 분기가 21.1~25.4 빠짐 없는지 확인
print(cafe_store.shape)
sorted(cafe_store["기준_년분기_코드"].unique())
```

```python
# interim에 저장
cafe_store.to_csv("../data/interim/cafe_store_2021_2025.csv", index=False, encoding="utf-8-sig")
```

# 2-1. 추정매출 누락 행 확인 ✅
점포 데이터 30,228행 중 8,656행은 추정매출이 없음 (매출 행은 전부 점포에 있음)
- 누락 행의 82.9%가 카페 0~3개 (매출 있는 행은 15.8%)
- → 카페가 적은 상권은 매출 추정에서 빠진다
- 단, 카페 20개 이상인데 매출 없는 행도 있음 → EDA에서 확인

```python
# 점포 수가 0인 상권이 점포수 데이터에만 있다? - 총 누락 8,656 중 826 개만 점포 수 0
print(len(cafe_store[cafe_store["점포_수"] == 0]))
```

```python
# 매출(cafe)이 조건이 될 때만 행을 만든다? / 추정매출 자료에만 있는 행 없음 확인 - 점포 수가 적은 상권이 누락된 경우가 대부분
check = pd.merge(cafe_store, cafe, on=["기준_년분기_코드", "서비스_업종_코드", "상권_코드"], how="left", indicator=True)
print(check["_merge"].value_counts())
missing = check[check["_merge"] == "left_only"]
missing["점포_수"].value_counts()
```

```python
# 추정매출 누락 데이터에서 점포 수 0~3개인 상권 82.9% / 정상 데이터에서 점포수 0~3개인 상권 15.8%
small = 0
for i in range(4):
    small = small + missing[missing["점포_수"] == i].shape[0]
print(round(small / missing.shape[0] * 100, 1))

both_rows = check[check["_merge"] == "both"]
small_both = 0
for i in range(4):
    small_both = small_both + both_rows[both_rows["점포_수"] == i].shape[0]
print(round(small_both / both_rows.shape[0] * 100, 1))
```
